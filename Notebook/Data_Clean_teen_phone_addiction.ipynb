Imported libraries together to be more streamlined and clean, same as the first cleanup notebook.

```python
df = pd.read_csv('../Data/teen_phone_addiction_dataset.csv')
df.head()
```

Running basic head, info, value counts, and shape to get information on what to do first for cleanup.

```python
df.info()
```

3000 rows, 25 columns, no missing values in any column. Bigger dataset than Teen_Mental_Health, but same clean starting point, no nulls to fix.

```python
df.shape
```

```python
df.describe()
```

Age runs 13 to 19, matching the first dataset. Academic_Performance here is on a 0 to 100 scale, not the 2.00 to 4.00 GPA scale used in the first dataset. Keeping that in mind for later, the two fields measure the same idea but are not directly comparable without converting one to match the other. Addiction_Level goes up to 10.0 and the mean sits close to 9, higher than the 25th percentile suggests, worth a closer look.

```python
df.duplicated().sum()
```

Zero duplicate rows.

```python
df['ID'].is_unique
```

ID column is unique for every row, good candidate for a primary key once renamed.

```python
df['Name'].duplicated().sum()
```

67 repeated names out of 3000 rows. Expected with randomly generated synthetic names, and Name is not useful for analysis anyway, so it gets dropped along with ID's raw form once user_id is built.

```python
df['Location'].nunique()
```

2726 unique values out of 3000 rows. Location is close to a free text field here, one value per person with almost no repeats, so it will not group into anything meaningful. Dropping it, same reasoning as dropping Name.

```python
df.drop(columns=['Name', 'Location'], inplace=True)
df.head()
```

Dropped Name and Location. Neither adds analytical value and both were closer to noise than a real category.

```python
df['Gender'].value_counts()
```

Male, Female, and Other are all close to 1000 rows each, an even three-way split. No fix needed here.

```python
df['School_Grade'].value_counts()
```

7th through 12th grade, each with roughly 480 to 530 students. Even spread, no missing grade levels.

```python
df['Phone_Usage_Purpose'].value_counts()
```

Five categories: Browsing, Other, Education, Social Media, Gaming. All within about 50 rows of each other.

```python
df['Parental_Control'].value_counts()
```

Binary field, 0 or 1, close to a 50/50 split.

```python
df['Addiction_Level'].value_counts().sort_index().tail(10)
```

This is the real finding in this dataset. 1524 of 3000 rows, just over half, sit at exactly 10.0, the maximum possible addiction score. The rest spread out across 1.0 to 9.9. That is a heavy ceiling effect, not an even synthetic spread like most of the other fields. Worth flagging clearly before this column drives any conclusions later.

```python
df.isna().sum()
```

Confirms no missing values remain after the column drops.

Renaming columns to snake_case so field names line up with the first dataset where the same idea is being measured, matching the intent already started in the trial notebook.

```python
df.rename(columns={
    'Age': 'age',
    'Gender': 'gender',
    'School_Grade': 'school_grade',
    'Daily_Usage_Hours': 'daily_usage_hours',
    'Sleep_Hours': 'sleep_hours',
    'Academic_Performance': 'academic_performance',
    'Social_Interactions': 'social_interactions',
    'Exercise_Hours': 'exercise_hours',
    'Anxiety_Level': 'anxiety_level',
    'Depression_Level': 'depression_level',
    'Self_Esteem': 'self_esteem',
    'Parental_Control': 'parental_control',
    'Screen_Time_Before_Bed': 'screen_time_before_bed',
    'Phone_Checks_Per_Day': 'phone_checks_per_day',
    'Apps_Used_Daily': 'apps_used_daily',
    'Time_on_Social_Media': 'time_on_social_media',
    'Time_on_Gaming': 'time_on_gaming',
    'Time_on_Education': 'time_on_education',
    'Phone_Usage_Purpose': 'phone_usage_purpose',
    'Family_Communication': 'family_communication',
    'Weekend_Usage_Hours': 'weekend_usage_hours',
    'Addiction_Level': 'addiction_level'
}, inplace=True)
df.head()
```

One naming note to remember for the analysis notebook: the first dataset has depression_label, a 0 or 1 flag. This dataset has depression_level, a 1 to 10 scale. Similar names, different meaning. Not combining these two fields directly.

```python
pd.options.display.float_format = '{:.2f}'.format
df.head()
```

Set float display to two decimal places, same formatting as the first cleanup notebook.

Building user_id from the ID column with a D2 prefix, same convention used for the first dataset's D1 prefix.

```python
df.rename(columns={'ID': 'user_id'}, inplace=True)
df['user_id'] = 'D2_' + df['user_id'].astype(str)
df.head()
```

```python
df['user_id'].is_unique
```

user_id stays unique after adding the prefix.

```python
df.shape
```

```python
df.info()
```

3000 rows, 23 columns after dropping Name and Location and adding user_id in place of the raw ID. No nulls, no duplicates.
