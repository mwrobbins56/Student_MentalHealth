This dataset has no user_id column, unlike both smartphone datasets. Adding one now so it can serve as a primary key in the database step. Using a D1 prefix to keep it distinct from the second dataset, same convention as the smartphone project.

```python
df.insert(0, 'user_id', ['D1_' + str(i).zfill(4) for i in range(1, len(df) + 1)])
df.head()
```
